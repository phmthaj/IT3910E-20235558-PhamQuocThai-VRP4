```python
import json
import random

def generate_vehicle_data_from_nodes(
    path_nodes_json,
    path_json_out="vehicles_standard.json",
    vehicles_per_depot=None,      # ví dụ [3,3] nếu có 2 depot
    vehicle_types=None,           # ví dụ ["type_1","type_2","type_3"]
    capacity_scales=None,         # ví dụ {"type_1":1.0, "type_2":1.5, "type_3":2.3}
    vehicle_costs=None,           # ví dụ {"type_1":1.0, "type_2":1.3, "type_3":1.8}
    ratio_to_vendor=0.5,          # tổng capacity mục tiêu = ratio * tổng cung vendor
    seed=42,
    equal_per_commodity=True      # NEW: True => mỗi xe có capacity các commodity bằng nhau
):
    random.seed(seed)

    # --- 1. Đọc dữ liệu node ---
    with open(path_nodes_json, "r", encoding="utf-8") as f:
        node_data = json.load(f)

    meta = node_data["meta"]
    n_depots = meta["n_depots"]
    commodities = meta["commodities"]
    K = len(commodities)

    # --- 2. Tính tổng cung từ vendor ---
    total_vendor = {c: 0.0 for c in commodities}
    for n in node_data["nodes"]:
        if n["type"] == "vendor":
            for c in commodities:
                total_vendor[c] += float(n["collected_goods_dict"][c])

    # --- 3. Tham số mặc định ---
    if vehicle_types is None:
        vehicle_types = ["type_1", "type_2", "type_3"]
    if capacity_scales is None:
        capacity_scales = {"type_1": 1.0, "type_2": 1.5, "type_3": 2.3}
    if vehicle_costs is None:
        vehicle_costs = {"type_1": 1.0, "type_2": 1.3, "type_3": 1.8}
    if vehicles_per_depot is None:
        vehicles_per_depot = [len(vehicle_types)] * n_depots

    assert len(vehicles_per_depot) == n_depots, \
        "⚠️ Độ dài vehicles_per_depot phải bằng số depot"

    total_vehicle = sum(vehicles_per_depot)

    # --- 4. Chuẩn bị tạo danh sách xe (để biết tổng scale) ---
    type_choices = []
    for depot_id, n_vehicle in enumerate(vehicles_per_depot):
        for _ in range(n_vehicle):
            type_choices.append(random.choice(vehicle_types))

    scales_list = [capacity_scales[t] for t in type_choices]
    sum_scales = sum(scales_list)
    scale_mean = sum_scales / total_vehicle if total_vehicle > 0 else 1.0

    # --- 5. Mục tiêu capacity ---
    total_vendor_all = sum(total_vendor.values())
    target_total_capacity_all = ratio_to_vendor * total_vendor_all

    vehicles = []
    v_id = 0

    if equal_per_commodity:
        # NEW: mỗi xe có capacity các commodity bằng nhau
        # base_scalar đảm bảo tổng capacity (gộp tất cả commodity & xe) đạt target
        if K == 0 or sum_scales == 0:
            base_scalar = 0.0
        else:
            base_scalar = target_total_capacity_all / (K * sum_scales)

        for depot_id, n_vehicle in enumerate(vehicles_per_depot):
            for _ in range(n_vehicle):
                type_name = type_choices[v_id]
                scale = capacity_scales[type_name]
                per_commodity_cap = round(base_scalar * scale, 1)
                cap_dict = {c: per_commodity_cap for c in commodities}
                vehicles.append({
                    "vehicle_id": f"D{depot_id}_V{v_id}",
                    "depot_id": depot_id,
                    "vehicle_type": type_name,
                    "capacity_dict": cap_dict
                })
                v_id += 1
    else:
        # Giữ logic cũ: capacity theo từng commodity, tỷ lệ theo cung vendor từng loại
        target_capacity = {c: total_vendor[c] * ratio_to_vendor for c in commodities}
        base_capacity = {
            c: (target_capacity[c] / (total_vehicle * scale_mean)) if total_vehicle > 0 else 0.0
            for c in commodities
        }
        for depot_id, n_vehicle in enumerate(vehicles_per_depot):
            for _ in range(n_vehicle):
                type_name = type_choices[v_id]
                scale = capacity_scales[type_name]
                cap_dict = {c: round(base_capacity[c] * scale, 1) for c in commodities}
                vehicles.append({
                    "vehicle_id": f"D{depot_id}_V{v_id}",
                    "depot_id": depot_id,
                    "vehicle_type": type_name,
                    "capacity_dict": cap_dict
                })
                v_id += 1

    data = {
        "meta": {
            "n_depots": n_depots,
            "n_vehicles": len(vehicles),
            "vehicle_types": vehicle_types,
            "vehicle_costs": vehicle_costs,
            "ratio_to_vendor": ratio_to_vendor,
            "capacity_mode": "equal_per_commodity" if equal_per_commodity else "proportional_by_vendor"
        },
        "vehicles": vehicles
    }

    with open(path_json_out, "w", encoding="utf-8") as f:
        json.dump(data, f, indent=2)

    # --- 6. Thống kê ---
    total_cap = {c: sum(v["capacity_dict"][c] for v in vehicles) for c in commodities}
    ratio_real_by_c = {
        k: (total_cap[k] / total_vendor[k] * 100) if total_vendor[k] > 0 else 0.0
        for k in commodities
    }
    ratio_real_total = (sum(total_cap.values()) / total_vendor_all * 100) if total_vendor_all > 0 else 0.0

    print(f"✅ Vehicle JSON saved to {path_json_out}")
    print(f"📦 Tổng xe mỗi depot: {vehicles_per_depot}")
    print("📦 Tổng cung vendor:", {k: round(v, 1) for k, v in total_vendor.items()})
    print("🚚 Tổng capacity xe:", {k: round(v, 1) for k, v in total_cap.items()})
    print("📉 Tỷ lệ capacity thực tế theo từng commodity:", {k: f'{v:.1f}%' for k, v in ratio_real_by_c.items()})
    print("📊 Tỷ lệ capacity thực tế (tổng):", f"{ratio_real_total:.1f}%")
    print("⚙️ Cấu hình scale:", capacity_scales)
    print("💰 Chi phí:", vehicle_costs)

```
