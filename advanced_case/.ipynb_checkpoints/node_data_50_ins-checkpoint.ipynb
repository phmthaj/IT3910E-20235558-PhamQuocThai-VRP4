```python
import xml.etree.ElementTree as ET
import pandas as pd
import numpy as np
import random
import json

def minutes(hh, mm): 
    return hh*60 + mm

def clamp(x, lo, hi):
    return max(lo, min(hi, x))

def convert_xml_to_json_vrp4_standard(
    path_xml,
    path_json_out,
    n_depots=3,
    n_vendors=None,
    n_markets=None,
    commodities=None,
    depot_capacity_factor=1.2,
    balance_factor=0.8,
    magnitude_factor=5.0,
    depot_selection="maxsum",
    # ---- NEW: cấu hình Time Window đơn giản ----
    depot_tw=(minutes(7,0),  minutes(19,0)),      # 07:00–19:00
    vendor_tw=(minutes(8,0), minutes(11,0)),      # 08:00–11:00 (sáng)
    market_tw=(minutes(13,0), minutes(17,0)),     # 13:00–17:00 (chiều)
    jitter_minutes=30,                            # nhiễu ±30' để data đỡ “phẳng”
    service_time_vendor=(10, 20),                 # phút, random trong khoảng
    service_time_market=(10, 25),
    service_time_depot=(0, 0),                    # thường là 0
    seed=42
):
    """
    Sinh file JSON chuẩn cho bài toán VRP4 + Time Window đơn giản:
      - Vendor: sáng (8–11h), Market: chiều (13–17h), Depot: 7–19h
      - Có jitter ±30' để đa dạng
      - Service time theo phút
    Depot chọn theo:
      - 'random'  : ngẫu nhiên
      - 'spread'  : xa depot gần nhất
      - 'maxsum'  : xa tổng thể (tổng khoảng cách đến tất cả depot trước)
    Các giá trị hàng hoá làm tròn 1 chữ số thập phân.
    """
    if commodities is None:
        commodities = ["food", "drink", "electronics", "household"]

    commodity_to_idx = {c: i for i, c in enumerate(commodities)}
    rng = random.Random(seed)
    np.random.seed(seed)

    # --- Load XML ---
    tree = ET.parse(path_xml)
    root = tree.getroot()

    nodes = []
    for node in root.findall("NodeP"):
        nodes.append({
            "id": int(node.find("id").text),
            "coord_x": float(node.find("CoordX").text),
            "coord_y": float(node.find("CoordY").text),
        })

    df = pd.DataFrame(nodes).sort_values("id").reset_index(drop=True)
    n_total = len(df)
    coords = df[["coord_x", "coord_y"]].values

    if n_vendors is None or n_markets is None:
        remaining = n_total - n_depots
        n_vendors = remaining // 2
        n_markets = remaining - n_vendors

    assert n_depots + n_vendors + n_markets <= n_total, (
        f"Sum depots+vendors+markets ({n_depots+n_vendors+n_markets}) > total nodes ({n_total})."
    )

    # --- Chọn depot ---
    if depot_selection == "random":
        depot_indices = rng.sample(range(n_total), n_depots)
    elif depot_selection == "spread":
        depot_indices = [rng.randrange(n_total)]
        while len(depot_indices) < n_depots:
            dists = np.min([np.linalg.norm(coords - coords[i], axis=1) for i in depot_indices], axis=0)
            for i in depot_indices: dists[i] = -1
            next_idx = int(np.argmax(dists)); depot_indices.append(next_idx)
    elif depot_selection == "maxsum":
        depot_indices = [rng.randrange(n_total)]
        while len(depot_indices) < n_depots:
            dists = np.sum([np.linalg.norm(coords - coords[i], axis=1) for i in depot_indices], axis=0)
            for i in depot_indices: dists[i] = -1
            next_idx = int(np.argmax(dists)); depot_indices.append(next_idx)
    else:
        raise ValueError("depot_selection must be 'random', 'spread', or 'maxsum'")

    depot_ids = df.loc[depot_indices, "id"].tolist()

    # --- Phân vendor/market ---
    remaining_ids = [nid for nid in df["id"] if nid not in depot_ids]
    rng.shuffle(remaining_ids)
    vendor_ids = remaining_ids[:n_vendors]
    market_ids = remaining_ids[n_vendors:n_vendors + n_markets]

    # --- Sinh dữ liệu hàng hoá ---
    vendor_data = [{c: round(rng.uniform(3, 12), 1) for c in commodities} for _ in range(n_vendors)]
    market_data = [{c: round(rng.uniform(2, 10), 1) for c in commodities} for _ in range(n_markets)]

    total_vendor = {c: sum(v[c] for v in vendor_data) for c in commodities}
    total_market = {c: sum(m[c] for m in market_data) for c in commodities}

    # Giới hạn cầu market ≤ cung * balance_factor
    for c in commodities:
        max_market_c = total_vendor[c] * balance_factor
        if total_market[c] > max_market_c and total_market[c] > 0:
            scale = max_market_c / total_market[c]
            for m in market_data:
                m[c] = round(m[c] * scale, 1)

    # Khuếch đại độ lớn bài toán
    for v in vendor_data:
        for c in commodities:
            v[c] = round(v[c] * magnitude_factor, 1)
    for m in market_data:
        for c in commodities:
            m[c] = round(m[c] * magnitude_factor, 1)

    total_vendor = {c: round(sum(v[c] for v in vendor_data), 1) for c in commodities}
    total_market = {c: round(sum(m[c] for m in market_data), 1) for c in commodities}

    # --- Sức chứa depot tổng & chia đều ---
    depot_capacity_total = {c: round(total_market[c] * 1.2, 1) for c in commodities}
    depot_capacity = {
        c: round((depot_capacity_total[c] / n_depots) * depot_capacity_factor, 1)
        for c in commodities
    }

    # --- Helper: tạo TW có jitter & đảm bảo hợp lệ ---
    def gen_tw(base_earliest, base_latest, jitter, min_slack=10):
        e = clamp(base_earliest + rng.randint(-jitter, +jitter), 0, 24*60)
        l = clamp(base_latest   + rng.randint(-jitter, +jitter), 0, 24*60)
        if l - e < min_slack:
            # đảm bảo còn tối thiểu min_slack phút mở cửa
            mid = (e + l) // 2
            e = max(0, mid - min_slack//2)
            l = min(24*60, e + min_slack)
        return {"earliest": int(e), "latest": int(l)}

    def rand_service(a_b_tuple):
        lo, hi = a_b_tuple
        return int(rng.randint(lo, hi))

    # --- Build node_data + gắn TW & service_time ---
    node_data = []
    v_idx, m_idx = 0, 0
    id_set = set(depot_ids) | set(vendor_ids) | set(market_ids)

    for _, row in df.iterrows():
        node_id = int(row["id"])
        if node_id not in id_set: 
            continue

        if node_id in depot_ids:
            cap_dict = depot_capacity
            cap_vec = [cap_dict[c] for c in commodities]
            node_data.append({
                "id": node_id,
                "type": "depot",
                "coord_x": row["coord_x"],
                "coord_y": row["coord_y"],
                "storage_capacity_dict": cap_dict,
                "storage_capacity_vec": cap_vec,
                "time_window": {
                    "earliest": int(depot_tw[0]),
                    "latest":   int(depot_tw[1]),
                },
                "service_time": rand_service(service_time_depot)
            })

        elif node_id in vendor_ids:
            v = vendor_data[v_idx]; v_idx += 1
            v_vec = [v[c] for c in commodities]
            node_data.append({
                "id": node_id,
                "type": "vendor",
                "coord_x": row["coord_x"],
                "coord_y": row["coord_y"],
                "collected_goods_dict": v,
                "collected_goods_vec": v_vec,
                "total_goods": round(sum(v.values()), 1),
                "time_window": gen_tw(vendor_tw[0], vendor_tw[1], jitter_minutes),
                "service_time": rand_service(service_time_vendor)
            })

        else:  # market
            m = market_data[m_idx]; m_idx += 1
            m_vec = [m[c] for c in commodities]
            node_data.append({
                "id": node_id,
                "type": "market",
                "coord_x": row["coord_x"],
                "coord_y": row["coord_y"],
                "order_dict": m,
                "order_vec": m_vec,
                "total_order": round(sum(m.values()), 1),
                "time_window": gen_tw(market_tw[0], market_tw[1], jitter_minutes),
                "service_time": rand_service(service_time_market)
            })

    data = {
        "meta": {
            "total_nodes": len(node_data),
            "n_depots": n_depots,
            "n_vendors": n_vendors,
            "n_markets": n_markets,
            "commodities": commodities,
            "commodity_to_idx": commodity_to_idx,
            "depot_capacity_factor": depot_capacity_factor,
            "balance_factor": balance_factor,
            "magnitude_factor": magnitude_factor,
            "depot_selection": depot_selection,
            "time_unit": "minute"                 # <--- quan trọng cho VRPTW
        },
        "nodes": node_data
    }

    with open(path_json_out, "w", encoding="utf-8") as f:
        json.dump(data, f, indent=2)

    total_depot_capacity = {c: round(depot_capacity[c] * n_depots, 1) for c in commodities}

    print(f"✅ JSON saved to {path_json_out}")
    print(f"🏭 Depot selection method: {depot_selection}")
    print(f"🏭 Depots: {depot_ids}")
    print(f"🏬 Vendors: {vendor_ids}")
    print(f"🏪 Markets: {market_ids}")
    print("📦 Tổng cung vendor:", total_vendor)
    print("🛒 Tổng cầu market:", total_market)
    print("🏢 Tổng sức chứa depot:", total_depot_capacity)
    print("📊 Depot capacity mỗi kho:", depot_capacity)
    print("⏱️ Time windows: depot=", depot_tw, ", vendor=", vendor_tw, ", market=", market_tw)






```

```python
import json
from pathlib import Path
import math

INPUT_PATH  = Path(r"C:\Users\thaip\Desktop\VRP4\data\JSON\osaba-50-instances\osaba_50_graph_matrix.json")
OUTPUT_PATH = Path(r"C:\Users\thaip\Desktop\VRP4\data\JSON\osaba-50-instances\osaba_50_time_matrix.json")
SPEED_M_PER_MIN = 1000   #

def add_travel_time_matrix():
    with open(INPUT_PATH, "r", encoding="utf-8") as f:
        data = json.load(f)

    distance_matrix = data["distance_matrix"]
    n = len(distance_matrix)

    travel_time_matrix = []
    for i in range(n):
        row = []
        for j in range(n):
            dist = distance_matrix[i][j]
            if isinstance(dist, (int, float)) and dist != float("inf"):
                # tính phút + ép về int (floor)
                time_min = int(dist / SPEED_M_PER_MIN)
                # để tránh 0 phút (vẫn cần >= 1 phút nếu có di chuyển)
                if time_min == 0 and dist > 0:
                    time_min = 1
                row.append(time_min)
            else:
                row.append("Infinity")
        travel_time_matrix.append(row)

    data["travel_time_matrix"] = travel_time_matrix
    data["time_unit"] = "minute"  # để AI khác đọc biết đơn vị

    with open(OUTPUT_PATH, "w", encoding="utf-8") as f:
        json.dump(data, f, indent=2, ensure_ascii=False)

    print(f"✅ DONE — saved to {OUTPUT_PATH}")

if __name__ == "__main__":
    add_travel_time_matrix()

```
