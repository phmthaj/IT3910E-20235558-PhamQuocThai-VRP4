```python
import xml.etree.ElementTree as ET
import pandas as pd
import random
import json


def convert_xml_to_json_vrp4_standard(
    path_xml,
    path_json_out,
    n_depots=3,
    n_vendors=None,
    n_markets=None,
    commodities=None,
    depot_capacity_factor=1.2,
    balance_factor=0.8,
    magnitude_factor=5.0,
    seed=42
):
    """
    Tạo file JSON chuẩn cho bài toán VRP4 có thêm commodity map.
    - Lưu cả dạng dict và dạng list (vector) theo thứ tự commodities.
    """

    if commodities is None:
        commodities = ["food", "drink", "electronics", "household"]

    commodity_to_idx = {c: i for i, c in enumerate(commodities)}

    random.seed(seed)

    tree = ET.parse(path_xml)
    root = tree.getroot()

    nodes = []
    for node in root.findall("NodeP"):
        nodes.append({
            "id": int(node.find("id").text),
            "coord_x": float(node.find("CoordX").text),
            "coord_y": float(node.find("CoordY").text),
        })

    df = pd.DataFrame(nodes).sort_values("id").reset_index(drop=True)
    n_total = len(df)

    if n_vendors is None or n_markets is None:
        remaining = n_total - n_depots
        n_vendors = remaining // 2
        n_markets = remaining - n_vendors

    assert n_depots + n_vendors + n_markets <= n_total, (
        f"Sum depots+vendors+markets ({n_depots+n_vendors+n_markets}) "
        f"> total nodes ({n_total})."
    )

    all_ids = df["id"].tolist()
    random.shuffle(all_ids)
    depot_ids = all_ids[:n_depots]
    vendor_ids = all_ids[n_depots:n_depots + n_vendors]
    market_ids = all_ids[n_depots + n_vendors:n_depots + n_vendors + n_markets]

    vendor_data = [{c: round(random.uniform(3, 12), 3) for c in commodities} for _ in range(n_vendors)]
    market_data = [{c: round(random.uniform(2, 10), 3) for c in commodities} for _ in range(n_markets)]

    total_vendor = {c: sum(v[c] for v in vendor_data) for c in commodities}
    total_market = {c: sum(m[c] for m in market_data) for c in commodities}

    for c in commodities:
        max_market_c = total_vendor[c] * balance_factor
        if total_market[c] > max_market_c:
            scale = max_market_c / total_market[c]
            for m in market_data:
                m[c] = round(m[c] * scale, 3)

    for v in vendor_data:
        for c in commodities:
            v[c] = round(v[c] * magnitude_factor, 2)
    for m in market_data:
        for c in commodities:
            m[c] = round(m[c] * magnitude_factor, 2)

    total_vendor = {c: sum(v[c] for v in vendor_data) for c in commodities}
    total_market = {c: sum(m[c] for m in market_data) for c in commodities}

    depot_capacity_total = {
        c: total_market[c] * 1.2 for c in commodities
    }
    depot_capacity = {
        c: round((depot_capacity_total[c] / n_depots) * depot_capacity_factor, 2)
        for c in commodities
    }

    node_data = []
    v_idx, m_idx = 0, 0
    id_set = set(depot_ids) | set(vendor_ids) | set(market_ids)

    for _, row in df.iterrows():
        node_id = int(row["id"])
        if node_id not in id_set:
            continue

        if node_id in depot_ids:
            cap_dict = depot_capacity
            cap_vec = [cap_dict[c] for c in commodities]
            node_data.append({
                "id": node_id,
                "type": "depot",
                "coord_x": row["coord_x"],
                "coord_y": row["coord_y"],
                "storage_capacity_dict": cap_dict,
                "storage_capacity_vec": cap_vec,
            })
        elif node_id in vendor_ids:
            v = vendor_data[v_idx]
            v_vec = [v[c] for c in commodities]
            node_data.append({
                "id": node_id,
                "type": "vendor",
                "coord_x": row["coord_x"],
                "coord_y": row["coord_y"],
                "collected_goods_dict": v,
                "collected_goods_vec": v_vec,
                "total_goods": round(sum(v.values()), 2)
            })
            v_idx += 1
        elif node_id in market_ids:
            m = market_data[m_idx]
            m_vec = [m[c] for c in commodities]
            node_data.append({
                "id": node_id,
                "type": "market",
                "coord_x": row["coord_x"],
                "coord_y": row["coord_y"],
                "order_dict": m,
                "order_vec": m_vec,
                "total_order": round(sum(m.values()), 2)
            })
            m_idx += 1

    data = {
        "meta": {
            "total_nodes": len(node_data),
            "n_depots": n_depots,
            "n_vendors": n_vendors,
            "n_markets": n_markets,
            "commodities": commodities,
            "commodity_to_idx": commodity_to_idx,
            "depot_capacity_factor": depot_capacity_factor,
            "balance_factor": balance_factor,
            "magnitude_factor": magnitude_factor
        },
        "nodes": node_data
    }

    with open(path_json_out, "w", encoding="utf-8") as f:
        json.dump(data, f, indent=2)

    total_depot_capacity = {c: depot_capacity[c] * n_depots for c in commodities}

    print(f"✅ JSON saved to {path_json_out}")
    print(f"🏭 Depots: {depot_ids}")
    print(f"🏬 Vendors: {vendor_ids}")
    print(f"🏪 Markets: {market_ids}")
    print("📦 Tổng cung vendor:", total_vendor)
    print("🛒 Tổng cầu market:", total_market)
    print("🏢 Tổng sức chứa depot:", total_depot_capacity)
    print("📊 Depot capacity mỗi kho:", depot_capacity)
    print("🧭 Commodity map:", commodity_to_idx)


convert_xml_to_json_vrp4_standard(
    path_xml=r"C:\Users\thaip\Desktop\VRP4\data\instances (english version)\Osaba_50_1_1.xml",
    path_json_out=r"C:\Users\thaip\Desktop\VRP4\data\JSON\osaba_50_standard.json",
    n_depots=4,
    n_vendors=18,
    n_markets=28,
    commodities=["food", "drink", "electronics", "household"],
    depot_capacity_factor=1.2,
    balance_factor=0.95,
    magnitude_factor=5.0,
    seed=42
)




```
