```python
import json
from pathlib import Path

BASE_DIR = Path(r"C:\Users\thaip\Desktop\VRP4\data\JSON\osaba-50-instances")
PATH_NODES_JSON   = BASE_DIR / "osaba_50_standard_v2.json"
PATH_OUT_CLUSTERS = BASE_DIR / "vendor_clusters_storage_optimizer.json"

GOODS_KEYS = ["food", "drink", "electronics", "household"]
LAMBDA_LOAD = 2.0
LOG_WARNINGS = True

def load_nodes_data(p):
    try:
        with open(p, "r", encoding="utf-8") as f:
            d = json.load(f)
    except:
        print(f"❌ Không tìm thấy file: {p}")
        return None
    n = d.get("nodes", [])
    return {"nodes": n, "depots": [x for x in n if x.get("type") == "depot"], "vendors": [x for x in n if x.get("type") == "vendor"]}

def get_vendor_demand_dict(v):
    od = v.get("collected_goods_dict", {}) or {}
    return {k: float(od.get(k, 0.0)) for k in GOODS_KEYS}

def manhattan_xy(a, b):
    return abs(float(a["coord_x"]) - float(b["coord_x"])) + abs(float(a["coord_y"]) - float(b["coord_y"]))

def build_initial_clusters(depots):
    res = []
    for d in depots:
        cap = d.get("storage_capacity_dict", {}) or {}
        res.append({
            "node_id": int(d["id"]),
            "center": {"coord_x": float(d["coord_x"]), "coord_y": float(d["coord_y"])},
            "capacity": {k: float(cap.get(k, 0.0)) for k in GOODS_KEYS},
            "current_mass": {k: 0.0 for k in GOODS_KEYS},
            "vendors": []
        })
    res.sort(key=lambda x: x["node_id"])
    return res

def apply_vendor_mass(cluster, vendor, sign):
    vd = get_vendor_demand_dict(vendor)
    for k in GOODS_KEYS:
        cluster["current_mass"][k] += sign * vd.get(k, 0.0)

def overflow_ratio_after_assign(cluster, vendor):
    eps = 1e-12
    vd = get_vendor_demand_dict(vendor)
    r = {}
    for k in GOODS_KEYS:
        cap = float(cluster["capacity"].get(k, 0.0))
        cur = float(cluster["current_mass"].get(k, 0.0))
        dem = float(vd.get(k, 0.0))
        if cap <= eps:
            r[k] = 1.0 if dem > 0 else 0.0
        else:
            over = max(0.0, (cur + dem) - cap)
            r[k] = over / cap
    return r

def can_assign_without_overflow(cluster, vendor):
    ratios = overflow_ratio_after_assign(cluster, vendor)
    return all(r <= 0.0 + 1e-12 for r in ratios.values())

def load_pressure(cluster):
    eps = 1e-12
    p = []
    for k in GOODS_KEYS:
        cap = float(cluster["capacity"].get(k, 0.0))
        cur = float(cluster["current_mass"].get(k, 0.0))
        if cap <= eps:
            p.append(1.0 if cur > 0 else 0.0)
        else:
            p.append(cur / cap)
    return max(p) if p else 0.0

def pick_depot_cost_based_hard(vendor, clusters, lam=LAMBDA_LOAD):
    feasible = []
    for i, c in enumerate(clusters):
        if can_assign_without_overflow(c, vendor):
            d = manhattan_xy(vendor, c["center"])
            p = load_pressure(c)
            cost = d * (1.0 + lam * p)
            feasible.append((cost, i))
    if feasible:
        feasible.sort(key=lambda x: x[0])
        return feasible[0][1], False
    order = sorted(range(len(clusters)), key=lambda i: manhattan_xy(vendor, clusters[i]["center"]))
    return order[0], True

def greedy_cost_based_hard(vendors, clusters, lam=LAMBDA_LOAD, log_warnings=True):
    unassigned = []
    forced = []
    sorted_v = sorted(vendors, key=lambda v: min(manhattan_xy(v, c["center"]) for c in clusters))
    for v in sorted_v:
        idx, was_forced = pick_depot_cost_based_hard(v, clusters, lam=lam)
        ratios = overflow_ratio_after_assign(clusters[idx], v)
        if was_forced:
            forced.append(int(v["id"]))
            if log_warnings:
                viol = {k: round(r * 100, 3) for k, r in ratios.items() if r > 0.0 + 1e-12}
                print(f"⚠ FORCED assign: Vendor {v['id']} không thể gán vào depot nào thỏa HARD-CAP; ép vào Depot {clusters[idx]['node_id']}. Vi phạm: {viol if viol else 'không có (capacity zero?)'}")
            clusters[idx]["vendors"].append(int(v["id"]))
            apply_vendor_mass(clusters[idx], v, +1)
            continue

        if idx is None:
            unassigned.append(int(v["id"]))
            if log_warnings:
                _ = get_vendor_demand_dict(v)
            continue
        clusters[idx]["vendors"].append(int(v["id"]))
        apply_vendor_mass(clusters[idx], v, +1)
    return unassigned, forced

def save_results(clusters, out_path):
    out = {
        "clusters": [
            {
                "depot_node_id": c["node_id"],
                "depot_storage_capacity": {k: round(float(v), 6) for k, v in c["capacity"].items()},
                "current_assigned_mass": {k: round(float(v), 6) for k, v in c["current_mass"].items()},
                "assigned_vendors": sorted(int(x) for x in c["vendors"]),
            }
            for c in sorted(clusters, key=lambda x: x["node_id"])
        ]
    }
    out_path.parent.mkdir(parents=True, exist_ok=True)
    with open(out_path, "w", encoding="utf-8") as f:
        json.dump(out, f, ensure_ascii=False, indent=2)
    print(f"\n✅ Đã lưu kết quả vào: {out_path}")

def print_cluster_summary(clusters, unassigned, forced):
    for c in sorted(clusters, key=lambda x: x["node_id"]):
        total_load = sum(c["current_mass"].values())
        total_cap  = sum(c["capacity"].values())
        print(f"\n--- Depot {c['node_id']} ---")
        print(f"  - Số vendor: {len(c['vendors'])}")
        print(f"  - Tổng tải / Sức chứa: {total_load:.2f} / {total_cap:.2f}")
        print("  - Theo loại hàng:")
        for k in GOODS_KEYS:
            print(f"    • {k:<12}: {c['current_mass'][k]:>8.2f} / {c['capacity'][k]:>8.2f}")
    if unassigned:
        print(f"\n❌ Tổng số vendor KHÔNG GÁN do HARD-CAP: {len(unassigned)}")
        print(f"   Danh sách: {sorted(unassigned)}")
    else:
        print("\n✅ Tất cả vendor đã được gán .")
    if forced:
        print(f"\n⚠ Tổng forced-assign (ép gán do không depot nào đủ): {len(forced)}")
        print(f"   Danh sách forced vendor: {sorted(forced)}")

def main():
    d = load_nodes_data(PATH_NODES_JSON)
    if not d:
        return
    depots = d["depots"]
    vendors = d["vendors"]
    if not depots:
        print("❌ Không có depot trong dữ liệu."); return
    if not vendors:
        print("⚠️  Không có vendor trong dữ liệu."); return
    clusters = build_initial_clusters(depots)
    print(f"▶️  Gán Vendor → Depot theo COST-BASED (λ={LAMBDA_LOAD}) với HARD-CAP (FORCE nếu tất cả depot đều vượt) …")
    unassigned, forced = greedy_cost_based_hard(vendors, clusters, lam=LAMBDA_LOAD, log_warnings=LOG_WARNINGS)
    print("\n📊 Tóm tắt:")
    print_cluster_summary(clusters, unassigned, forced)
    save_results(clusters, PATH_OUT_CLUSTERS)

if __name__ == "__main__":
    main()






```

```python
import json
import random
from math import ceil, floor, exp
from pathlib import Path
from dataclasses import dataclass, field
from typing import List, Dict, Tuple, Callable, Any

SEED = 42
random.seed(SEED)

BASE_DIR = Path(r"C:\Users\thaip\Desktop\VRP4\data\JSON\osaba-50-instances")
PATH_CLUSTERS = BASE_DIR / "Cluster" / "vendor_clusters_storage_optimizer.json"
PATH_VEHICLES = BASE_DIR / "vehicles_standard.json"
PATH_DIST     = BASE_DIR / "osaba_50_time_matrix.json"
PATH_NODES    = BASE_DIR / "osaba_50_standard_v2.json"
PATH_OUT      = BASE_DIR / "routes_alns_pickup.json"

GOODS = ["food", "drink", "electronics", "household"]

SCALE = 1000
DIST_SCALE = 1.0
CHUNK_SAFETY = 0.95
DEPOT_NODE_TO_INTERNAL = {26: 0, 38: 1, 40: 2, 41: 3}
HUGE = 1_000_000_000

# ALNS meta-params
SEGMENT_LENGTH = 50          # số vòng giữa hai lần cập nhật trọng số
RHO = 0.6                    # hệ số làm mịn trọng số (0.5~0.8)
REWARD_GLOBAL_BEST = 6.0     # ψ1: đạt nghiệm tốt nhất lịch sử
REWARD_IMPROVED     = 3.0    # ψ2: nghiệm mới tốt hơn nghiệm hiện tại
REWARD_ACCEPTED     = 1.0    # ψ3: nghiệm xấu nhưng vẫn được chấp nhận
EARLY_STOP_PATIENCE = 500    # dừng sớm nếu không cải thiện best_cost sau N vòng


def load_json(path: Path) -> Any:
    with open(path, "r", encoding="utf-8") as f:
        return json.load(f)

def build_id_to_idx(nodes_json) -> Dict[int, int]:
    return {int(n["id"]): idx for idx, n in enumerate(nodes_json["nodes"])}

def dist_accessor(distance_matrix, id_to_idx) -> Callable[[int, int], float]:
    """
    Trả về hàm d(a,b) đọc từ ma trận; cạnh 'Infinity'/None => HUGE.
    """
    n = len(distance_matrix)
    def d(a: int, b: int) -> float:
        ia = id_to_idx.get(int(a), None)
        ib = id_to_idx.get(int(b), None)
        if ia is None or ib is None: return HUGE
        if not (0 <= ia < n and 0 <= ib < n): return HUGE
        val = distance_matrix[ia][ib]
        if isinstance(val, str) and val.lower() == "infinity": return HUGE
        if val is None: return HUGE
        return float(val)
    return d

def get_vendor_demand(nodes_json) -> Dict[int, Dict[str, float]]:
    out = {}
    for n in nodes_json["nodes"]:
        if n.get("type") != "vendor":
            continue
        vid = int(n["id"])
        gdict = (
            n.get("collected_goods_dict")
            or n.get("collected_goods")
            or n.get("order_dict")
            or n.get("goods_dict")
            or {}
        )
        out[vid] = {g: float(gdict.get(g, 0.0)) for g in GOODS}
    return out

def vehicles_of_depot(vehicles_json, depot_internal: int) -> List[dict]:
    vs = [v for v in vehicles_json["vehicles"] if int(v["depot_id"]) == depot_internal]
    if not vs:
        raise RuntimeError(f"Không có xe cho depot_id={depot_internal}")
    return vs

def vehicle_capacity_vec(v: dict) -> List[int]:
    return [int(round(float(v["capacity_dict"][g]) * SCALE)) for g in GOODS]

def vehicle_cost_per_km(v: dict, vehicle_costs_table: Dict[str, float]) -> float:
    t = v.get("vehicle_type", "type_1")
    return float(vehicle_costs_table.get(t, 1.0))

def sub_vector_inplace(a: List[int], b: List[int]) -> None:
    for i in range(len(a)): a[i] -= b[i]

def add_vector_inplace(a: List[int], b: List[int]) -> None:
    for i in range(len(a)): a[i] += b[i]

def leq_vec(a: List[int], b: List[int]) -> bool:
    return all(a[i] <= b[i] for i in range(len(a)))


def integer_split_scaled_safe(vendors: List[int], demand_map: Dict[int, Dict[str, float]], min_cap_scaled: int):
    """
    Chia mỗi vendor thành các 'chunk' theo TỔNG khối lượng (sum over GOODS),
    với mỗi chunk <= CHUNK_SAFETY * min_cap_scaled.
    Việc kiểm tra capacity vector thực hiện khi chèn.
    """
    if min_cap_scaled <= 0:
        min_cap_scaled = 1
    limit = int(max(1, CHUNK_SAFETY * min_cap_scaled))
    chunks_nodes: List[int] = []
    chunks_qty_scaled: List[int] = []
    chunk_detail: List[None] = []

    for vid in vendors:
        total = float(sum(demand_map.get(vid, {}).get(g, 0.0) for g in GOODS))
        total_scaled = int(round(total * SCALE))
        if total_scaled <= 0:
            continue

        need = ceil(total_scaled / limit)
        base = total_scaled // need
        rem  = total_scaled % need
        for i in range(need):
            take = base + (1 if i < rem else 0)
            take = min(take, limit)
            take = max(1, take)
            chunks_nodes.append(vid)
            chunks_qty_scaled.append(int(take))
            chunk_detail.append(None)

    return chunks_nodes, chunks_qty_scaled, chunk_detail

def allocate_chunk_vector(vendor_remaining_vec: List[int], qty_scaled: int) -> List[int]:
    """
    Phân bổ qty_scaled vào vector commodities theo tỉ lệ phần còn lại của vendor.
    Clamp để không vượt mức còn lại từng commodity.
    """
    total_rem = sum(max(0, x) for x in vendor_remaining_vec)
    if total_rem <= 0:
        return [0]*len(GOODS)

    alloc = []
    for rem_i in vendor_remaining_vec:
        portion = (rem_i / total_rem) if total_rem > 0 else 0.0
        part = int(round(qty_scaled * portion))
        alloc.append(part)

    # điều chỉnh sai số do làm tròn
    diff = qty_scaled - sum(alloc)
    i = 0
    while diff != 0 and i < len(alloc):
        if diff > 0:
            alloc[i] += 1; diff -= 1
        else:
            take = min(1, alloc[i])
            alloc[i] -= take; diff += take
        i = (i+1) % len(alloc)

    # clamp theo lượng còn lại
    for i in range(len(alloc)):
        if alloc[i] > vendor_remaining_vec[i]:
            alloc[i] = vendor_remaining_vec[i]
    return alloc



@dataclass
class Route:
    depot: int
    vehicle_idx: int
    veh_cost: float
    cap_vec: List[int]
    path: List[int] = field(default_factory=list)         # [depot, ..., depot]
    load_vec: List[int] = field(default_factory=list)     # tổng tải cuối route (vector)
    assigned_chunks: List[Dict[str, Any]] = field(default_factory=list)  # [{"vendor_id": vid, "alloc": vec}]
    dist_cache: float = 0.0

    @staticmethod
    def new(depot_node_id: int, vehicle_idx: int, veh_cost_per_km: float, cap_vec: List[int]) -> "Route":
        return Route(
            depot=int(depot_node_id),
            vehicle_idx=vehicle_idx,
            veh_cost=float(veh_cost_per_km),
            cap_vec=cap_vec[:],
            path=[int(depot_node_id), int(depot_node_id)],
            load_vec=[0]*len(GOODS),
            assigned_chunks=[],
            dist_cache=0.0
        )

    def clone_shallow(self) -> "Route":
        return Route(
            depot=self.depot,
            vehicle_idx=self.vehicle_idx,
            veh_cost=self.veh_cost,
            cap_vec=self.cap_vec[:],
            path=self.path[:],
            load_vec=self.load_vec[:],
            assigned_chunks=[{"vendor_id": c["vendor_id"], "alloc": c["alloc"][:]} for c in self.assigned_chunks],
            dist_cache=self.dist_cache
        )

@dataclass
class Solution:
    routes: List[Route]
    unassigned: set
    vendor_remaining: Dict[int, List[int]]
    cost: float = 0.0

    def clone(self) -> "Solution":
        return Solution(
            routes=[r.clone_shallow() for r in self.routes],
            unassigned=set(self.unassigned),
            vendor_remaining={vid: vec[:] for vid, vec in self.vendor_remaining.items()},
            cost=self.cost
        )


# =====================
# DISTANCE, COST & INSERT
# =====================
def compute_route_distance(route: Route, dist_fn: Callable[[int,int], float]) -> float:
    total = 0.0
    p = route.path
    for i in range(len(p)-1):
        d = dist_fn(p[i], p[i+1])
        if d >= HUGE:
            return float(HUGE)
        total += d
    return total

def recompute_all_cost(sol: Solution, dist_fn: Callable[[int,int], float]) -> float:
    total = 0.0
    for r in sol.routes:
        d = compute_route_distance(r, dist_fn)
        if d >= HUGE:
            return float(HUGE)
        total += d * r.veh_cost
    sol.cost = total
    return total

def delta_insert_edge(route: Route, vendor_id: int, dist_fn: Callable[[int,int], float], insert_pos: int) -> float:
    """
    Δdistance khi CHÈN vendor_id vào trước vị trí insert_pos (giữa (u,v) = (path[pos-1], path[pos])).
    """
    p = route.path
    u = p[insert_pos-1]
    v = p[insert_pos]
    duv = dist_fn(u, v)
    dux = dist_fn(u, vendor_id)
    dxv = dist_fn(vendor_id, v)
    if dux >= HUGE or dxv >= HUGE or duv >= HUGE:
        return float(HUGE)
    return (dux + dxv - duv)

def apply_insert_at(route: Route, vendor_id: int, alloc_vec: List[int], delta_d: float, insert_pos: int) -> None:
    route.path.insert(insert_pos, vendor_id)
    add_vector_inplace(route.load_vec, alloc_vec)
    route.assigned_chunks.append({"vendor_id": vendor_id, "alloc": alloc_vec[:]})
    route.dist_cache += float(delta_d)

def feasible_add_chunk_to_route(route: Route, alloc_vec: List[int]) -> bool:
    """
    Pickup-only: kiểm tra tổng tải cuối route không vượt capacity vector.
    """
    new_load = [route.load_vec[i] + alloc_vec[i] for i in range(len(GOODS))]
    return leq_vec(new_load, route.cap_vec)


# ==============================
# INITIAL FEASIBLE CONSTRUCTION
# ==============================
def build_initial_solution_for_cluster(cluster, vehicles_json, vehicle_costs, dist_fn, demand_map, min_cap_scaled) -> Solution:
    depot_node = int(cluster["depot_node_id"])
    depot_internal = DEPOT_NODE_TO_INTERNAL[depot_node]
    vendors = [int(v) for v in cluster["assigned_vendors"]]

    vehicles = vehicles_of_depot(vehicles_json, depot_internal)
    if not vehicles:
        raise RuntimeError(f"No vehicles for depot {depot_node}")

    # khởi tạo các route rỗng (mỗi xe một route trống)
    routes: List[Route] = []
    for vi, v in enumerate(vehicles):
        cap_vec = vehicle_capacity_vec(v)
        veh_cost = vehicle_cost_per_km(v, vehicles_json["meta"]["vehicle_costs"])
        routes.append(Route.new(depot_node, vi, veh_cost, cap_vec))

    per_caps_total = [sum(vehicle_capacity_vec(v)) for v in vehicles]
    min_cap = int(max(1, floor(min(per_caps_total))))
    chunk_nodes, chunk_qty_scaled, _ = integer_split_scaled_safe(vendors, demand_map, min_cap_scaled=min_cap)

    # vendor_remaining khởi tạo bằng demand ban đầu
    vendor_remaining: Dict[int, List[int]] = {}
    for vid in vendors:
        vec = [int(round(float(demand_map.get(vid, {}).get(g, 0.0))*SCALE)) for g in GOODS]
        vendor_remaining[vid] = vec

    unassigned: List[int] = []
    # chèn từng chunk bằng best-position insertion
    for (vid, qty_scaled) in zip(chunk_nodes, chunk_qty_scaled):
        alloc = allocate_chunk_vector(vendor_remaining[vid], qty_scaled)
        if sum(alloc) <= 0:
            continue

        best = None
        for r in routes:
            if not feasible_add_chunk_to_route(r, alloc):
                continue
            for pos in range(1, len(r.path)):  # thử mọi cạnh
                delta_d = delta_insert_edge(r, vid, dist_fn, pos)
                if delta_d >= HUGE:
                    continue
                delta_cost = delta_d * r.veh_cost
                if (best is None) or (delta_cost < best[0]):
                    best = (delta_cost, r, delta_d, pos)

        if best is not None:
            _, r, delta_d, pos = best
            apply_insert_at(r, vid, alloc, delta_d, pos)
            sub_vector_inplace(vendor_remaining[vid], alloc)
        else:
            # mở route mới từ xe rẻ nhất/km
            best_vi = min(range(len(vehicles)),
                          key=lambda vi: vehicle_cost_per_km(vehicles[vi], vehicles_json["meta"]["vehicle_costs"]))
            cap_vec = vehicle_capacity_vec(vehicles[best_vi])
            veh_cost = vehicle_cost_per_km(vehicles[best_vi], vehicles_json["meta"]["vehicle_costs"])
            new_r = Route.new(depot_node, best_vi, veh_cost, cap_vec)

            if feasible_add_chunk_to_route(new_r, alloc):
                delta_d = delta_insert_edge(new_r, vid, dist_fn, insert_pos=1)  # chỉ có (depot->depot)
                if delta_d < HUGE:
                    apply_insert_at(new_r, vid, alloc, delta_d, insert_pos=1)
                    sub_vector_inplace(vendor_remaining[vid], alloc)
                    routes.append(new_r)
                else:
                    unassigned.append(vid)
            else:
                unassigned.append(vid)

    sol = Solution(routes, set(unassigned), vendor_remaining)
    recompute_all_cost(sol, dist_fn)
    return sol



def list_all_visits(sol: Solution) -> List[Tuple[int,int,int]]:
    """Trả list (ri, pos, vid) cho mọi vendor visit trong các route (bỏ depot)."""
    locs: List[Tuple[int,int,int]] = []
    for ri, r in enumerate(sol.routes):
        for pos in range(1, len(r.path)-1):
            vid = r.path[pos]
            if vid != r.depot:
                locs.append((ri, pos, vid))
    return locs

def remove_visit(sol: Solution, ri: int, pos: int) -> Tuple[int, List[int]]:
    """
    Gỡ 1 visit tại route ri, vị trí pos.
    Lấy 1 chunk tương ứng trong assigned_chunks (vendor_id=vid) để trả hàng về vendor_remaining.
    """
    r = sol.routes[ri]
    vid = r.path[pos]
    alloc = None
    for i in reversed(range(len(r.assigned_chunks))):
        if r.assigned_chunks[i]["vendor_id"] == vid:
            alloc = r.assigned_chunks.pop(i)["alloc"]
            break
    if alloc is None:
        alloc = [0]*len(GOODS)

    # trả tải về vendor_remaining & xóa node
    for j in range(len(GOODS)):
        r.load_vec[j] -= alloc[j]
        sol.vendor_remaining[vid][j] += alloc[j]

    r.path.pop(pos)
    return (vid, alloc)

def random_removal(sol: Solution, dist_fn, remove_ratio=0.15):
    locs = list_all_visits(sol)
    if not locs:
        return []
    k = max(1, int(len(locs)*remove_ratio))
    picked = random.sample(locs, k)
    picked.sort(key=lambda t: (t[0], t[1]), reverse=True)

    removed = []
    for (ri, pos, vid) in picked:
        removed.append(remove_visit(sol, ri, pos))
    return removed

def worst_removal(sol: Solution, dist_fn, remove_ratio=0.15):
    locs = list_all_visits(sol)
    if not locs:
        return []
    contribs = []
    for (ri, pos, vid) in locs:
        r = sol.routes[ri]
        u = r.path[pos-1]; x = r.path[pos]; v = r.path[pos+1]
        duv = dist_fn(u, v)
        c = (dist_fn(u, x) + dist_fn(x, v) - duv) * r.veh_cost
        contribs.append(((ri, pos, vid), c))
    contribs.sort(key=lambda t: t[1], reverse=True)
    k = max(1, int(len(locs)*remove_ratio))
    picked_idx = [it[0] for it in contribs[:k]]
    picked_idx.sort(key=lambda t: (t[0], t[1]), reverse=True)

    removed = []
    for (ri, pos, vid) in picked_idx:
        removed.append(remove_visit(sol, ri, pos))
    return removed

def shaw_removal(sol: Solution, dist_fn, remove_ratio=0.15):
    """
    Shaw removal (chỉ dựa khoảng cách):
    - Chọn 1 seed visit
    - Loại thêm các visit "gần seed" nhất theo khoảng cách (distance(seed, vid))
    -> Phù hợp mô hình đã CHUNK: không dùng demand nữa.
    """
    locs = list_all_visits(sol)
    if not locs:
        return []
    k = max(1, int(len(locs)*remove_ratio))

    seed_ri, seed_pos, seed_vid = random.choice(locs)

    cand = []
    for (ri, pos, vid) in locs:
        if (ri, pos, vid) == (seed_ri, seed_pos, seed_vid):
            continue
        dist = dist_fn(seed_vid, vid)
        cand.append(((ri, pos, vid), dist))

    cand.sort(key=lambda t: t[1])  # nhỏ = gần hơn
    picked_idx = [(seed_ri, seed_pos, seed_vid)] + [c[0] for c in cand[:k-1]]
    picked_idx.sort(key=lambda t: (t[0], t[1]), reverse=True)

    removed = []
    for (ri, pos, vid) in picked_idx:
        removed.append(remove_visit(sol, ri, pos))
    return removed



def greedy_best_position_insertion(sol: Solution, removed_list, dist_fn) -> bool:
    """
    Chèn từng (vid, alloc) vào vị trí có Δcost nhỏ nhất (trên mọi route, mọi cạnh).
    Nếu không chèn được, mở route mới theo xe rẻ nhất/km của solution.
    """
    for (vid, alloc_vec) in removed_list:
        if sum(alloc_vec) <= 0:
            continue

        best = None
        for r in sol.routes:
            if not feasible_add_chunk_to_route(r, alloc_vec):
                continue
            for pos in range(1, len(r.path)):
                delta_d = delta_insert_edge(r, vid, dist_fn, pos)
                if delta_d >= HUGE:
                    continue
                delta_cost = delta_d * r.veh_cost
                if (best is None) or (delta_cost < best[0]):
                    best = (delta_cost, r, delta_d, pos)

        if best is None:
            if not sol.routes:
                return False
            ref = min(sol.routes, key=lambda rr: rr.veh_cost)
            new_r = Route.new(ref.depot, ref.vehicle_idx, ref.veh_cost, ref.cap_vec)
            if not feasible_add_chunk_to_route(new_r, alloc_vec):
                return False
            delta_d = delta_insert_edge(new_r, vid, dist_fn, insert_pos=1)
            if delta_d >= HUGE:
                return False
            apply_insert_at(new_r, vid, alloc_vec, delta_d, insert_pos=1)
            sol.routes.append(new_r)
        else:
            _, r, delta_d, pos = best
            apply_insert_at(r, vid, alloc_vec, delta_d, pos)

        for j in range(len(GOODS)):
            sol.vendor_remaining[vid][j] -= alloc_vec[j]
    return True

def regret2_best_position_insertion(sol: Solution, removed_list, dist_fn) -> bool:
    """
    Với mỗi (vid, alloc), xét 2 vị trí tốt nhất (Δcost) trên toàn bộ các route.
    Ưu tiên chèn những vendor có regret = cost2 - cost1 lớn nhất trước.
    """
    candidates = []
    for (vid, alloc_vec) in removed_list:
        if sum(alloc_vec) <= 0:
            continue
        two_best = []
        for r in sol.routes:
            if not feasible_add_chunk_to_route(r, alloc_vec):
                continue
            for pos in range(1, len(r.path)):
                delta_d = delta_insert_edge(r, vid, dist_fn, pos)
                if delta_d >= HUGE:
                    continue
                delta_cost = delta_d * r.veh_cost
                two_best.append((delta_cost, r, delta_d, pos))
        two_best.sort(key=lambda x: x[0])

        if len(two_best) == 0:
            # không có chỗ chèn trên routes hiện hữu → mở route mới
            if not sol.routes:
                return False
            ref = min(sol.routes, key=lambda rr: rr.veh_cost)
            new_r = Route.new(ref.depot, ref.vehicle_idx, ref.veh_cost, ref.cap_vec)
            if not feasible_add_chunk_to_route(new_r, alloc_vec):
                return False
            delta_d = delta_insert_edge(new_r, vid, dist_fn, insert_pos=1)
            if delta_d >= HUGE:
                return False
            candidates.append((vid, alloc_vec, [(delta_d*new_r.veh_cost, new_r, delta_d, 1)]))
        else:
            candidates.append((vid, alloc_vec, two_best[:2]))

    def regret_score(lst):
        if len(lst) == 1:
            return 1e9
        return lst[1][0] - lst[0][0]

    candidates.sort(key=lambda item: regret_score(item[2]), reverse=True)

    for (vid, alloc_vec, best_list) in candidates:
        best_choice = best_list[0]
        _, r, delta_d, pos = best_choice
        if not feasible_add_chunk_to_route(r, alloc_vec):
            ref = min(sol.routes, key=lambda rr: rr.veh_cost) if sol.routes else None
            if ref is None:
                return False
            new_r = Route.new(ref.depot, ref.vehicle_idx, ref.veh_cost, ref.cap_vec)
            if not feasible_add_chunk_to_route(new_r, alloc_vec):
                return False
            delta_d2 = delta_insert_edge(new_r, vid, dist_fn, insert_pos=1)
            if delta_d2 >= HUGE:
                return False
            apply_insert_at(new_r, vid, alloc_vec, delta_d2, insert_pos=1)
            sol.routes.append(new_r)
        else:
            apply_insert_at(r, vid, alloc_vec, delta_d, pos)

        for j in range(len(GOODS)):
            sol.vendor_remaining[vid][j] -= alloc_vec[j]
    return True



def roulette_select(weights: List[float]) -> int:
    s = sum(weights)
    if s <= 0:
        return int(random.random() * len(weights))
    threshold = random.random() * s
    acc = 0.0
    for i, w in enumerate(weights):
        acc += w
        if acc >= threshold:
            return i
    return len(weights)-1

def update_weights(weights: List[float], scores: List[float], counts: List[int], rho: float = RHO) -> None:
    """
    w_i <- (1-rho)*w_i + rho*(score_i / max(1, count_i))
    Reset scores & counts sau khi cập nhật.
    """
    for i in range(len(weights)):
        avg = scores[i] / max(1, counts[i])
        weights[i] = (1 - rho)*weights[i] + rho*avg
        scores[i] = 0.0
        counts[i] = 0


# ============================
# ALNS LOOP PER CLUSTER (MAIN)
# ============================
def alns_optimize_cluster(cluster, vehicles_json, dist_fn, demand_map,
                          max_iter=5000, destroy_ratio=0.15, cooling=0.998):
    depot_node = int(cluster["depot_node_id"])
    depot_internal = DEPOT_NODE_TO_INTERNAL[depot_node]
    vehicles = vehicles_of_depot(vehicles_json, depot_internal)
    per_caps_total = [sum(vehicle_capacity_vec(v)) for v in vehicles]
    if not per_caps_total or min(per_caps_total) <= 0:
        return {"depot_node": depot_node, "routes": [], "distance_cost_sum": 0.0}

    min_cap = int(max(1, floor(min(per_caps_total))))
    curr = build_initial_solution_for_cluster(
        cluster, vehicles_json, vehicles_json["meta"]["vehicle_costs"],
        dist_fn, demand_map, min_cap
    )
    best = curr.clone()
    best_cost = recompute_all_cost(best, dist_fn)
    curr_cost = best_cost

    # Operators registry
    destroy_ops = [
        ("random_removal", lambda s: random_removal(s, dist_fn, remove_ratio=destroy_ratio)),
        ("worst_removal",  lambda s: worst_removal(s, dist_fn, remove_ratio=destroy_ratio)),
        ("shaw_removal",   lambda s: shaw_removal(s, dist_fn, remove_ratio=destroy_ratio)),
    ]
    repair_ops = [
        ("greedy",   lambda s, rem: greedy_best_position_insertion(s, rem, dist_fn)),
        ("regret-2", lambda s, rem: regret2_best_position_insertion(s, rem, dist_fn)),
    ]
    w_d = [1.0]*len(destroy_ops)
    w_r = [1.0]*len(repair_ops)
    scr_d = [0.0]*len(destroy_ops)
    cnt_d = [0]*len(destroy_ops)
    scr_r = [0.0]*len(repair_ops)
    cnt_r = [0]*len(repair_ops)

    T = max(1.0, best_cost * 0.05)
    iter_since_last_weight_update = 0
    best_iter = 0  # for early stopping

    for it in range(max_iter):
        di = roulette_select(w_d)
        ri = roulette_select(w_r)
        destroy_name, destroy_fn = destroy_ops[di]
        repair_name,  repair_fn  = repair_ops[ri]

        trial = curr.clone()

        # DESTROY
        removed = destroy_fn(trial)
        if not removed:
            continue

        # REPAIR
        ok = repair_fn(trial, removed)
        if not ok:
            continue

        new_cost = recompute_all_cost(trial, dist_fn)

        # scoring
        improved_curr = new_cost < curr_cost
        improved_best = new_cost < best_cost
        if improved_best:
            reward = REWARD_GLOBAL_BEST
        elif improved_curr:
            reward = REWARD_IMPROVED
        else:
            reward = REWARD_ACCEPTED

        scr_d[di] += reward; cnt_d[di] += 1
        scr_r[ri] += reward; cnt_r[ri] += 1

        # SA acceptance
        delta = new_cost - curr_cost
        accept = (delta < 0) or (random.random() < exp(-delta / max(T, 1e-9)))
        if accept:
            curr = trial
            curr_cost = new_cost
            if improved_best:
                best = trial.clone()
                best_cost = new_cost
                best_iter = it

        # Cooling
        T *= cooling

        # Adaptive weights update (per segment)
        iter_since_last_weight_update += 1
        if iter_since_last_weight_update >= SEGMENT_LENGTH:
            update_weights(w_d, scr_d, cnt_d, rho=RHO)
            update_weights(w_r, scr_r, cnt_r, rho=RHO)
            iter_since_last_weight_update = 0

        # Early stopping (nếu quá lâu không cải thiện best)
        if it - best_iter >= EARLY_STOP_PATIENCE:
            break

    # Xuất kết quả
    routes_out, total = [], 0.0
    for r in best.routes:
        d = compute_route_distance(r, dist_fn)
        if d >= HUGE:
            continue
        dist_cost = d * r.veh_cost
        routes_out.append({
            "vehicle_id": f"depot{depot_internal}_veh{r.vehicle_idx}",
            "vehicle_cost_per_km": r.veh_cost,
            "path_node_ids": r.path[:],
            "distance_raw": float(d),
            "route_cost": float(dist_cost)
        })
        total += dist_cost

    return {"depot_node": depot_node, "routes": routes_out, "distance_cost_sum": total}


def main():
    random.seed(SEED)

    clusters  = load_json(PATH_CLUSTERS)
    vehicles  = load_json(PATH_VEHICLES)
    dist_json = load_json(PATH_DIST)
    dist_mat  = dist_json["distance_matrix"]
    nodes     = load_json(PATH_NODES)
    demand    = get_vendor_demand(nodes)
    id2idx    = build_id_to_idx(nodes)
    dist_fn   = dist_accessor(dist_mat, id2idx)

    total_cost = 0.0
    results = []

    for c in clusters["clusters"]:
        res = alns_optimize_cluster(
            c,
            vehicles_json=vehicles,
            dist_fn=dist_fn,
            demand_map=demand,
            max_iter=1000,
            destroy_ratio=0.18,
            cooling=0.998
        )
        results.append(res)
        total_cost += float(res.get("distance_cost_sum", 0.0))

    out = {
        "objective": "sum(distance * vehicle_type_cost) over routes",
        "total_cost": total_cost,
        "distance_unit": ("km" if DIST_SCALE == 1000.0 else "raw"),
        "seed": SEED,
        "clusters": results
    }
    with open(PATH_OUT, "w", encoding="utf-8") as f:
        json.dump(out, f, ensure_ascii=False, indent=2)

    print(f"Saved ALNS pickup result to: {PATH_OUT}")
    print(f"Total cost: {total_cost:.2f} ({'km*cost' if DIST_SCALE==1000.0 else 'raw*cost'})")


if __name__ == "__main__":
    main()


```

```python
import json
from pathlib import Path
import math

try:
    BASE_DIR = Path(r"C:\Users\thaip\Desktop\VRP4\data\JSON\osaba-50-instances")
    PATH_NODES_JSON = BASE_DIR / "osaba_50_standard_v2.json"
    PATH_VENDOR_CLUSTERS = BASE_DIR / "vendor_clusters_storage_optimizer.json"
except Exception as e:
    print(f"Lỗi khi thiết lập đường dẫn: {e}")


def load_data_from_path(filepath: Path):
    """Tải dữ liệu từ một tệp JSON."""
    with open(filepath, 'r', encoding='utf-8') as f:
        return json.load(f)

def manhattan_distance(node1, node2):
    """Tính khoảng cách Manhattan giữa hai điểm (node)."""
    return abs(node1['coord_x'] - node2['coord_x']) + abs(node1['coord_y'] - node2['coord_y'])

def capacitated_distance_clustering():
    """
    Hàm chính thực hiện thuật toán phân cụm kết hợp khoảng cách và ràng buộc tồn kho.
    """
    print("Bắt đầu thuật toán phân cụm theo khoảng cách Manhattan và tồn kho...")
    
    try:
        vendor_clusters = load_data_from_path(PATH_VENDOR_CLUSTERS)
        osaba_data = load_data_from_path(PATH_NODES_JSON)
    except FileNotFoundError as e:
        print(f"\n[LỖI] Không tìm thấy tệp: {e.filename}")
        return

    # --- Bước 1: Chuẩn bị dữ liệu ---
    commodities = osaba_data['meta']['commodities']
    all_nodes = {node['id']: node for node in osaba_data['nodes']}
    
    markets = [node for node in osaba_data['nodes'] if node['type'] == 'market']
    depots_nodes = {node['id']: node for node in osaba_data['nodes'] if node['type'] == 'depot'}

    # Khởi tạo kho hàng
    initial_depots = {c['depot_node_id']: c['current_assigned_mass'].copy() for c in vendor_clusters['clusters']}
    depots_stock = {c['depot_node_id']: c['current_assigned_mass'].copy() for c in vendor_clusters['clusters']}

    # --- Bước 2: Gán ưu tiên (Full Assignment) ---
    unassigned_markets = markets.copy()
    temp_assignments = {depot_id: [] for depot_id in depots_stock}
    
    while True:
        best_assignment = None # Lưu dạng (distance, market, depot_id)
        
        # Tìm tất cả các cặp (thị trường, depot) hợp lệ
        possible_assignments = []
        for market in unassigned_markets:
            for depot_id, stock in depots_stock.items():
                # Kiểm tra ràng buộc tồn kho
                if all(stock.get(item, 0) >= market['order_dict'].get(item, 0) for item in market['order_dict']):
                    depot_node = depots_nodes[depot_id]
                    distance = manhattan_distance(market, depot_node)
                    possible_assignments.append((distance, market, depot_id))
        
        # Nếu không còn cặp nào có thể gán toàn bộ, thoát vòng lặp
        if not possible_assignments:
            break
            
        # Chọn cặp tốt nhất (khoảng cách ngắn nhất)
        best_assignment = min(possible_assignments, key=lambda x: x[0])
        _, market_to_assign, depot_id_to_assign = best_assignment
        
        # Thực hiện gán
        order = market_to_assign['order_dict']
        temp_assignments[depot_id_to_assign].append({'market_id': market_to_assign['id'], 'order_split': order})
        
        # Cập nhật tồn kho
        for item, amount in order.items():
            depots_stock[depot_id_to_assign][item] -= amount
            
        # Cập nhật danh sách thị trường chưa gán
        unassigned_markets = [m for m in unassigned_markets if m['id'] != market_to_assign['id']]

    print(f"Hoàn thành gán ưu tiên. Còn lại {len(unassigned_markets)} thị trường cần chia nhỏ đơn hàng.")

    # --- Bước 3: Xử lý chia nhỏ đơn hàng cho các thị trường còn lại ---
    for market in unassigned_markets:
        remaining_order = market['order_dict'].copy()
        
        # Sắp xếp các depot theo khoảng cách từ gần đến xa
        depots_sorted_by_distance = sorted(
            depots_stock.keys(),
            key=lambda depot_id: manhattan_distance(market, depots_nodes[depot_id])
        )
        
        for depot_id in depots_sorted_by_distance:
            stock = depots_stock[depot_id]
            split_for_depot = {}
            
            for item, needed in remaining_order.items():
                if needed > 0 and stock.get(item, 0) > 0:
                    amount_to_take = min(needed, stock[item])
                    split_for_depot[item] = amount_to_take
                    stock[item] -= amount_to_take
                    remaining_order[item] -= amount_to_take
            
            if split_for_depot:
                temp_assignments[depot_id].append({'market_id': market['id'], 'order_split': split_for_depot})

    print("\n==================================================")
    print("📊 TÓM TẮT THÔNG TIN PHÂN CỤM THEO DEPOT")
    print("==================================================")
    
    output_data = {"clusters": []}
    for depot_id, initial_stock in initial_depots.items():
        remaining_goods = depots_stock[depot_id]
        assigned_mass = {item: round(initial_stock[item] - remaining_goods[item], 2) for item in commodities}
        assigned_markets_ids = sorted(list(set(a['market_id'] for a in temp_assignments[depot_id])))

        print(f"\n--- Cụm Depot ID: {depot_id} ---")
        print(f"  Các thị trường được phục vụ: {assigned_markets_ids}")
        print(f"  {'Loại hàng':<15} | {'Tồn kho ban đầu':<20} | {'Đã vận chuyển':<20}")
        print("-" * 60)
        for item in commodities:
            print(f"  {item.capitalize():<15} | {initial_stock.get(item, 0):<20.2f} | {assigned_mass.get(item, 0):<20.2f}")

        cluster_obj = {
            "depot_node_id": depot_id,
            "remaining_goods": {item: round(amount, 2) for item, amount in remaining_goods.items()},
            "assigned_mass": assigned_mass,
            "assigned_markets": assigned_markets_ids
        }
        output_data["clusters"].append(cluster_obj)

    output_filepath = BASE_DIR / 'market_clusters_storage_optimizer.json'
    try:
        with open(output_filepath, 'w', encoding='utf-8') as f:
            json.dump(output_data, f, indent=2)
        print("\n--------------------------------------------------")
        print("✅ Hoàn thành! Quá trình phân cụm đã thành công.")
        print(f"📂 Kết quả đã được lưu vào tệp: {output_filepath}")
        print("--------------------------------------------------")
    except Exception as e:
        print(f"\n[LỖI] Không thể ghi kết quả ra tệp: {e}")


if __name__ == "__main__":
    capacitated_distance_clustering()
```

```python

import json
import math
import random
from math import ceil, floor
from pathlib import Path
from typing import Dict, List



SEED = 42
random.seed(SEED)

BASE_DIR = Path(r"C:\Users\thaip\Desktop\VRP4\data\JSON\osaba-50-instances")
PATH_CLUSTERS = BASE_DIR/"Cluster"/ "market_clusters_storage_optimizer.json"   
PATH_VEHICLES = BASE_DIR / "vehicles_standard.json"
PATH_DIST     = BASE_DIR / "osaba_50_time_matrix.json"
PATH_NODES    = BASE_DIR / "osaba_50_standard_v2.json"

OUT_DIR  = BASE_DIR / "Sol_ALNS"
OUT_DIR.mkdir(parents=True, exist_ok=True)
PATH_OUT = OUT_DIR / "routes_alns_delivery.JSON"

# Tên commodity phải khớp trong JSON
GOODS = ["food", "drink", "electronics", "household"]

# Scale demand → integer
SCALE = 1000

# Khoảng cách: nếu cần đổi đơn vị thì chỉnh DIST_SCALE
DIST_SCALE = 1.0

# Chunking an toàn theo min-cap xe
CHUNK_SAFETY = 0.95

# Map node-id (bên ngoài) → depot internal index (khớp dataset của bạn)
DEPOT_NODE_TO_INTERNAL = {26: 0, 38: 1, 40: 2, 41: 3}

# Đại diện edge cấm
HUGE = 1_000_000_000



def load_json(path: Path):
    with open(path, "r", encoding="utf-8") as f:
        return json.load(f)

def build_id_to_idx(nodes_json):
    return {int(n["id"]): idx for idx, n in enumerate(nodes_json["nodes"])}

def dist_accessor(distance_matrix, id_to_idx):
    n = len(distance_matrix)
    def d(a, b):
        ia = id_to_idx.get(int(a), None)
        ib = id_to_idx.get(int(b), None)
        if ia is None or ib is None: return HUGE
        if not (0 <= ia < n and 0 <= ib < n): return HUGE
        val = distance_matrix[ia][ib]
        if isinstance(val, str) and val.lower() == "infinity": return HUGE
        if val is None: return HUGE
        return float(val)
    return d

def get_market_demand(nodes_json):
    """Market dùng 'order_dict' theo xác nhận của bạn."""
    out = {}
    for n in nodes_json["nodes"]:
        if n.get("type") != "market":
            continue
        mid = int(n["id"])
        gdict = (
            n.get("order_dict")
            or n.get("demand_dict")
            or {}
        )
        out[mid] = {g: float(gdict.get(g, 0.0)) for g in GOODS}
    return out

def vehicles_of_depot(vehicles_json, depot_internal):
    vs = [v for v in vehicles_json["vehicles"] if int(v["depot_id"]) == depot_internal]
    if not vs:
        raise RuntimeError(f"Không có xe cho depot_id={depot_internal}")
    return vs

def vehicle_capacity_vec(v):
    return [int(round(float(v["capacity_dict"][g]) * SCALE)) for g in GOODS]

def vehicle_cost_per_km(v, vehicle_costs_table):
    t = v.get("vehicle_type", "type_1")
    return float(vehicle_costs_table.get(t, 1.0))

def integer_split_scaled_safe_custom(nodes_list, demand_map, min_cap_scaled):
    """
    Chia demand theo 'tổng mass' để tạo chunk an toàn (mỗi chunk <= CHUNK_SAFETY * min_cap_scaled).
    Route sẽ kiểm tra capacity vector khi chèn.
    """
    if min_cap_scaled <= 0:
        min_cap_scaled = 1
    limit = int(max(1, CHUNK_SAFETY * min_cap_scaled))
    chunk_nodes = []
    chunk_qty_scaled = []

    for nid in nodes_list:
        total = float(sum(demand_map.get(nid, {}).get(g, 0.0) for g in GOODS))
        total_scaled = int(round(total * SCALE))
        if total_scaled <= 0:
            continue

        need = ceil(total_scaled / limit)
        base = total_scaled // need
        rem  = total_scaled % need
        for i in range(need):
            take = base + (1 if i < rem else 0)
            take = min(take, limit)
            take = max(1, take)
            chunk_nodes.append(nid)
            chunk_qty_scaled.append(int(take))
    return chunk_nodes, chunk_qty_scaled


class Route:
    """
    Một trip của 1 vehicle:
      depot -> seq(market-chunks) -> depot
    load_vec: tải hiện tại (scaled) theo commodity (chất lên từ depot để giao; kiểm tra ≤ cap_vec)
    """
    def __init__(self, depot_node_id, vehicle_idx, veh_cost_per_km, cap_vec):
        self.depot = int(depot_node_id)
        self.vehicle_idx = vehicle_idx
        self.veh_cost = float(veh_cost_per_km)
        self.cap_vec = cap_vec[:]
        self.path = [self.depot, self.depot]
        self.load_vec = [0]*len(GOODS)
        self.assigned_chunks: List[Dict] = []  # {"market_id": mid, "alloc": [..]}
        self.dist_cache = 0.0

    def clone_shallow(self):
        r = Route(self.depot, self.vehicle_idx, self.veh_cost, self.cap_vec)
        r.path = self.path[:]
        r.load_vec = self.load_vec[:]
        r.assigned_chunks = [dict(market_id=x["market_id"], alloc=x["alloc"][:]) for x in self.assigned_chunks]
        r.dist_cache = self.dist_cache
        return r

class Solution:
    """
    Lời giải của 1 cluster (depot):
      - routes: list[Route]
      - market_remaining: mid -> vector remaining (scaled)
    """
    def __init__(self, routes, market_remaining):
        self.routes = routes
        self.market_remaining = market_remaining
        self.cost = 0.0

    def clone(self):
        routes_copy = [r.clone_shallow() for r in self.routes]
        rem_copy = {mid: vec[:] for mid, vec in self.market_remaining.items()}
        s = Solution(routes_copy, rem_copy)
        s.cost = self.cost
        return s


def compute_route_distance(route: Route, dist_fn):
    total = 0.0
    p = route.path
    for i in range(len(p)-1):
        d = dist_fn(p[i], p[i+1])
        if d >= HUGE:
            return float(HUGE)
        total += d
    return total

def recompute_all_cost(sol: Solution, dist_fn):
    total = 0.0
    for r in sol.routes:
        d = compute_route_distance(r, dist_fn)
        if d >= HUGE:
            return float(HUGE)
        total += d * r.veh_cost
    sol.cost = total
    return total


def allocate_chunk_vector(remaining_vec, qty_scaled):
    """
    Phân bổ qty_scaled vào vector commodity theo tỉ lệ remaining.
    """
    total_rem = sum(max(0, x) for x in remaining_vec)
    if total_rem <= 0:
        return [0]*len(GOODS)

    alloc = []
    for rem_i in remaining_vec:
        portion = (rem_i / total_rem) if total_rem > 0 else 0.0
        part = int(round(qty_scaled * portion))
        alloc.append(part)

    diff = qty_scaled - sum(alloc)
    i = 0
    while diff != 0 and i < len(alloc):
        if diff > 0:
            alloc[i] += 1; diff -= 1
        else:
            take = min(1, alloc[i])
            alloc[i] -= take; diff += take
        i = (i+1) % len(alloc)

    # không vượt remaining
    for i in range(len(alloc)):
        if alloc[i] > remaining_vec[i]:
            alloc[i] = remaining_vec[i]
    return alloc

def sub_vector_inplace(a, b):
    for i in range(len(a)): a[i] -= b[i]

def add_vector_inplace(a, b):
    for i in range(len(a)): a[i] += b[i]

def leq_vec(a, b):
    return all(a[i] <= b[i] for i in range(len(a)))

def feasible_add_chunk_to_route_tail(route: Route, alloc_vec):
    # giao hàng: chất lên xe tại depot → constraint là tổng khối lượng cần giao của route ≤ cap_vec
    new_load = [route.load_vec[i] + alloc_vec[i] for i in range(len(GOODS))]
    return leq_vec(new_load, route.cap_vec)

def delta_insert_tail_distance(route: Route, market_id, dist_fn):
    p = route.path
    insert_pos = len(p)-1
    u = p[insert_pos-1]
    v = p[insert_pos]    # depot
    duv = dist_fn(u, v)
    dux = dist_fn(u, market_id)
    dxv = dist_fn(market_id, v)
    if dux >= HUGE or dxv >= HUGE or duv >= HUGE:
        return float(HUGE)
    return (dux + dxv - duv)

def apply_insert_tail(route: Route, market_id, alloc_vec, delta_d):
    route.path.insert(len(route.path)-1, market_id)
    add_vector_inplace(route.load_vec, alloc_vec)
    route.assigned_chunks.append({"market_id": market_id, "alloc": alloc_vec[:]})
    route.dist_cache += float(delta_d)

def build_initial_solution_for_cluster(cluster, vehicles_json, vehicle_costs, dist_fn, demand_map):
    depot_node = int(cluster["depot_node_id"])
    depot_internal = DEPOT_NODE_TO_INTERNAL[depot_node]
    markets = [int(m) for m in cluster["assigned_markets"]]

    vehicles = vehicles_of_depot(vehicles_json, depot_internal)
    if not vehicles:
        raise RuntimeError(f"No vehicles for depot {depot_node}")

    routes = []
    for vi, v in enumerate(vehicles):
        cap_vec = vehicle_capacity_vec(v)
        veh_cost = vehicle_cost_per_km(v, vehicle_costs)
        routes.append(Route(depot_node, vi, veh_cost, cap_vec))

    per_caps_total = [sum(vehicle_capacity_vec(v)) for v in vehicles]
    min_cap = int(max(1, floor(min(per_caps_total))))
    chunk_nodes, chunk_qty_scaled = integer_split_scaled_safe_custom(markets, demand_map, min_cap_scaled=min_cap)

    market_remaining = {}
    for mid in markets:
        vec = [int(round(float(demand_map.get(mid, {}).get(g, 0.0)) * SCALE)) for g in GOODS]
        market_remaining[mid] = vec

    for mid, qty_scaled in zip(chunk_nodes, chunk_qty_scaled):
        alloc = allocate_chunk_vector(market_remaining[mid], qty_scaled)
        if sum(alloc) <= 0:
            continue

        placed = False
        for r in routes:
            if feasible_add_chunk_to_route_tail(r, alloc):
                delta_d = delta_insert_tail_distance(r, mid, dist_fn)
                if delta_d >= HUGE:  # forbidden edges
                    continue
                apply_insert_tail(r, mid, alloc, delta_d)
                sub_vector_inplace(market_remaining[mid], alloc)
                placed = True
                break

        if not placed:
            best_vi = min(range(len(vehicles)),
                          key=lambda vi: vehicle_cost_per_km(vehicles[vi], vehicle_costs))
            cap_vec = vehicle_capacity_vec(vehicles[best_vi])
            veh_cost = vehicle_cost_per_km(vehicles[best_vi], vehicle_costs)
            new_r = Route(depot_node, best_vi, veh_cost, cap_vec)
            if feasible_add_chunk_to_route_tail(new_r, alloc):
                delta_d = delta_insert_tail_distance(new_r, mid, dist_fn)
                if delta_d < HUGE:
                    apply_insert_tail(new_r, mid, alloc, delta_d)
                    sub_vector_inplace(market_remaining[mid], alloc)
                    routes.append(new_r)

    sol = Solution(routes, market_remaining)
    recompute_all_cost(sol, dist_fn)
    return sol


def list_all_market_positions(sol: Solution):
    locs = []
    for ri, r in enumerate(sol.routes):
        for pi in range(1, len(r.path)-1):
            nid = r.path[pi]
            if nid != r.depot:
                locs.append((ri, pi))
    return locs

def random_removal(sol: Solution, dist_fn, remove_ratio=0.18):
    locs = list_all_market_positions(sol)
    if not locs:
        return []

    k = max(1, int(len(locs)*remove_ratio))
    picked = random.sample(locs, k=k)
    picked.sort(key=lambda x: (-x[0], -x[1]))  # xóa từ cuối

    removed = []
    for (ri, pi) in picked:
        r = sol.routes[ri]
        mid = r.path[pi]
        alloc = None
        for i in reversed(range(len(r.assigned_chunks))):
            if r.assigned_chunks[i]["market_id"] == mid:
                alloc = r.assigned_chunks.pop(i)["alloc"]
                break
        if alloc is None:
            alloc = [0]*len(GOODS)
        for j in range(len(GOODS)):
            r.load_vec[j] -= alloc[j]
            sol.market_remaining[mid][j] += alloc[j]
        r.path.pop(pi)
        r.dist_cache = compute_route_distance(r, dist_fn)
        removed.append((mid, alloc))

    return removed

def greedy_repair(sol: Solution, removed_list, dist_fn):
    for (mid, alloc_vec) in removed_list:
        best = None
        for r in sol.routes:
            if not feasible_add_chunk_to_route_tail(r, alloc_vec):
                continue
            delta_d = delta_insert_tail_distance(r, mid, dist_fn)
            if delta_d >= HUGE:
                continue
            delta_cost = delta_d * r.veh_cost
            if (best is None) or (delta_cost < best[0]):
                best = (delta_cost, r, delta_d)
        if best is None:
            if not sol.routes:
                return False
            ref_route = min(sol.routes, key=lambda rr: rr.veh_cost)
            new_r = Route(ref_route.depot, ref_route.vehicle_idx, ref_route.veh_cost, ref_route.cap_vec)
            delta_d = delta_insert_tail_distance(new_r, mid, dist_fn)
            if delta_d >= HUGE or not feasible_add_chunk_to_route_tail(new_r, alloc_vec):
                return False
            apply_insert_tail(new_r, mid, alloc_vec, delta_d)
            sol.routes.append(new_r)
            for j in range(len(GOODS)):
                sol.market_remaining[mid][j] -= alloc_vec[j]
        else:
            _, r, delta_d = best
            apply_insert_tail(r, mid, alloc_vec, delta_d)
            for j in range(len(GOODS)):
                sol.market_remaining[mid][j] -= alloc_vec[j]
    return True

def alns_optimize_cluster(cluster, vehicles_json, dist_fn, demand_map,
                          max_iter=1500, destroy_ratio=0.18, cooling=0.998):
    depot_node = int(cluster["depot_node_id"])
    depot_internal = DEPOT_NODE_TO_INTERNAL[depot_node]
    vehicles = vehicles_of_depot(vehicles_json, depot_internal)
    per_caps_total = [sum(vehicle_capacity_vec(v)) for v in vehicles]
    if not per_caps_total or min(per_caps_total) <= 0:
        return {"depot_node": depot_node, "routes": [], "distance_cost_sum": 0.0}

    init_sol = build_initial_solution_for_cluster(
        cluster, vehicles_json, vehicles_json["meta"]["vehicle_costs"], dist_fn, demand_map
    )
    best = init_sol.clone()
    best_cost = recompute_all_cost(best, dist_fn)

    curr = init_sol.clone()
    curr_cost = best_cost

    T = max(1.0, best_cost * 0.05)
    for _ in range(max_iter):
        removed = random_removal(curr, dist_fn, remove_ratio=destroy_ratio)
        if not removed:
            continue
        ok = greedy_repair(curr, removed, dist_fn)
        if not ok:
            curr = best.clone()
            curr_cost = best_cost
            continue

        new_cost = recompute_all_cost(curr, dist_fn)
        if new_cost < best_cost:
            best = curr.clone()
            best_cost = new_cost

        delta = new_cost - curr_cost
        if (delta < 0) or (random.random() < math.exp(-delta / max(T, 1e-9))):
            curr_cost = new_cost
        else:
            curr = best.clone()
            curr_cost = best_cost

        T *= cooling

    routes_out, total = [], 0.0
    for r in best.routes:
        d = compute_route_distance(r, dist_fn)
        if d >= HUGE:
            continue
        dist_cost = d * r.veh_cost
        routes_out.append({
            "vehicle_id": f"depot{depot_internal}_veh{r.vehicle_idx}",
            "vehicle_cost_per_km": r.veh_cost,
            "path_node_ids": r.path[:],
            "distance_raw": float(d),
            "route_cost": float(dist_cost)
        })
        total += dist_cost

    return {"depot_node": depot_node, "routes": routes_out, "distance_cost_sum": total}



def main():
    random.seed(SEED)

    clusters  = load_json(PATH_CLUSTERS)
    vehicles  = load_json(PATH_VEHICLES)
    dist_json = load_json(PATH_DIST)
    dist_mat  = dist_json["distance_matrix"]
    nodes     = load_json(PATH_NODES)
    demand    = get_market_demand(nodes)     
    id2idx    = build_id_to_idx(nodes)
    dist_fn   = dist_accessor(dist_mat, id2idx)

    total_cost = 0.0
    results = []

    for c in clusters["clusters"]:
        res = alns_optimize_cluster(
            c,
            vehicles_json=vehicles,
            dist_fn=dist_fn,
            demand_map=demand,
            max_iter=1500,
            destroy_ratio=0.18,
            cooling=0.998
        )
        results.append(res)
        total_cost += float(res.get("distance_cost_sum", 0.0))

    out = {
        "phase": "delivery",
        "objective": "sum(distance * vehicle_type_cost) over routes",
        "total_cost": total_cost,
        "distance_unit": ("km" if DIST_SCALE == 1000.0 else "raw"),
        "seed": SEED,
        "clusters": results
    }
    with open(PATH_OUT, "w", encoding="utf-8") as f:
        json.dump(out, f, ensure_ascii=False, indent=2)

    print(f"✅ Saved ALNS delivery result to: {PATH_OUT}")
    print(f"🧮 Total cost: {total_cost:.2f} ({'km*cost' if DIST_SCALE==1000.0 else 'raw*cost'})")

if __name__ == "__main__":
    main()
```

```python
import pandas as pd
import json
import matplotlib.pyplot as plt
from matplotlib.patches import Wedge
from matplotlib.lines import Line2D
from pathlib import Path
from collections import defaultdict

try:
    BASE_DIR = Path(r"C:\Users\thaip\Desktop\VRP4\data\JSON\osaba-50-instances")
    
    NODES_FILE = BASE_DIR / "osaba_50_standard_v2.json"
    
    CLUSTERS_FILE = BASE_DIR / "market_clusters_storage_optimizer.json"
    
    OUTPUT_IMAGE_FILE = BASE_DIR / "market_clusters_visualization.png"

except Exception as e:
    print(f"Lỗi khi thiết lập đường dẫn: {e}")

CLUSTER_COLORS = {26: 'red', 38: 'blue', 40: 'green', 41: 'orange'}



def visualize_market_clusters():
    print("Bắt đầu quá trình vẽ biểu đồ phân cụm...")
    
    try:
        with open(NODES_FILE, 'r', encoding='utf-8') as f:
            nodes_data = json.load(f)['nodes']
        nodes_df = pd.DataFrame(nodes_data)[['id', 'type', 'coord_x', 'coord_y']].copy()
        nodes_df['id'] = nodes_df['id'].astype(int)

        with open(CLUSTERS_FILE, 'r', encoding='utf-8') as f:
            clusters = json.load(f).get('clusters', [])
    except FileNotFoundError as e:
        print(f"\n[LỖI] Không tìm thấy tệp: {e.filename}")
        print("Vui lòng đảm bảo bạn đã chạy thuật toán phân cụm trước và tệp kết quả tồn tại.")
        return
    except Exception as e:
        print(f"\n[LỖI] Đã xảy ra sự cố khi đọc tệp: {e}")
        return

    market_to_depots_map = defaultdict(list)
    for cluster in clusters:
        depot_id = int(cluster['depot_node_id'])
        for market_id in cluster.get('assigned_markets', []):
            market_to_depots_map[int(market_id)].append(depot_id)

    multi_assigned_markets = {
        m_id: depots for m_id, depots in market_to_depots_map.items() if len(depots) > 1
    }
    print(f"Đã tìm thấy {len(multi_assigned_markets)} thị trường được chia nhỏ.")

    depots_df = nodes_df[nodes_df['type'] == 'depot'].copy()
    depots_df['color'] = depots_df['id'].map(CLUSTER_COLORS)
    
    markets_df = nodes_df[nodes_df['type'] == 'market'].copy()
    
    normal_markets_df = markets_df[~markets_df['id'].isin(multi_assigned_markets.keys())].copy()
    normal_market_colors = {m_id: CLUSTER_COLORS[depots[0]] for m_id, depots in market_to_depots_map.items() if len(depots) == 1}
    normal_markets_df['color'] = normal_markets_df['id'].map(normal_market_colors)
    
    plt.style.use("ggplot")
    fig, ax = plt.subplots(figsize=(12, 10))

    for m_id, depots_assigned in multi_assigned_markets.items():
        m_row = nodes_df[nodes_df['id'] == m_id].iloc[0]
        x, y = m_row['coord_x'], m_row['coord_y']
        
        num_splits = len(depots_assigned)
        angle_per_split = 360 / num_splits
        
        for i, depot_id in enumerate(sorted(depots_assigned)): 
            start_angle = i * angle_per_split
            end_angle = (i + 1) * angle_per_split
            color = CLUSTER_COLORS.get(depot_id, 'gray')
            wedge = Wedge((x, y), 120, start_angle, end_angle, facecolor=color, edgecolor='black', lw=0.7)
            ax.add_patch(wedge)

        ax.annotate(f"{m_id}", (x, y), textcoords="offset points",
                    xytext=(0, -15), ha='center', fontsize=7, color='black', weight='bold')

    ax.scatter(normal_markets_df['coord_x'], normal_markets_df['coord_y'],
               c=normal_markets_df['color'], marker='^', s=70, alpha=0.9, edgecolors='black', lw=0.5)
    for _, r in normal_markets_df.iterrows():
        ax.annotate(f"{int(r['id'])}", (r['coord_x'], r['coord_y']),
                    textcoords="offset points", xytext=(0, 8), ha='center', fontsize=7)

    # Vẽ các kho (hình vuông)
    ax.scatter(depots_df['coord_x'], depots_df['coord_y'],
               c=depots_df['color'], marker='s', s=250, edgecolors='black', linewidth=1.5, zorder=5)
    for _, r in depots_df.iterrows():
        ax.annotate(f"Kho {int(r['id'])}", (r['coord_x'], r['coord_y']),
                    textcoords="offset points", xytext=(0, -18), ha='center',
                    color='black', fontsize=10, fontweight='bold')

    legend_elems = [
        Line2D([0], [0], marker='s', color='w', label=f'Kho {k}',
               markerfacecolor=v, markersize=12, markeredgecolor='black')
        for k, v in CLUSTER_COLORS.items()
    ]
    legend_elems.append(Line2D([0], [0], marker='^', color='w', label='Market (1 Kho)',
                               markerfacecolor='gray', markersize=10, markeredgecolor='black'))
    legend_elems.append(Line2D([0], [0], marker='o', color='gray', label='Market (chia nhỏ)',
                               markerfacecolor='white', markeredgecolor='black', markersize=10))

    ax.legend(handles=legend_elems, title='Chú giải', loc='upper right', fontsize=10)
    ax.set_title("Phân cụm Market theo Kho (Kết hợp Khoảng cách & Tồn kho)", fontsize=16)
    ax.set_xlabel("Tọa độ X", fontsize=12)
    ax.set_ylabel("Tọa độ Y", fontsize=12)
    fig.tight_layout()

    try:
        plt.savefig(OUTPUT_IMAGE_FILE, dpi=300)
        print("\n--------------------------------------------------")
        print("✅ Hoàn thành! Biểu đồ đã được tạo thành công.")
        print(f"📂 Kết quả đã được lưu vào tệp: {OUTPUT_IMAGE_FILE}")
        print("--------------------------------------------------")
    except Exception as e:
        print(f"\n[LỖI] Không thể lưu tệp hình ảnh: {e}")


if __name__ == "__main__":
    visualize_market_clusters()
```
