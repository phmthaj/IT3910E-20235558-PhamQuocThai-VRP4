```python
import optuna
from ALNS_advanced import parse_vrp_file, alns

DATA_PATH = "data/instances/Loggi-n401-k23.vrp"
CACHED_DATA = parse_vrp_file(DATA_PATH)

def objective(trial):
    reward_improve = trial.suggest_float("reward_improve", 5.0, 20.0)
    reward_accept = trial.suggest_float("reward_accept", 1.0, 5.0)
    
    weight_decay = trial.suggest_float("weight_decay", 0.5, 0.95)
    weight_learning = trial.suggest_float("weight_learning", 0.05, 0.5) 
    
    cooling = trial.suggest_float("cooling", 0.99, 0.9999) 
    T0_factor = trial.suggest_float("T0_factor", 0.01, 0.1)

    dist = CACHED_DATA["dist"]
    depot = CACHED_DATA["depot"]
    demands = CACHED_DATA["demands"]
    cap = CACHED_DATA["capacity"]

    _, best_cost, _ = alns(
        dist, depot, demands, cap,
        iters=300, 
        seed=42, 
        time_limit=30, 
        reward_improve=reward_improve,
        reward_accept=reward_accept,
        weight_decay=weight_decay,
        weight_learning=weight_learning,
        T0_factor=T0_factor,
        cooling=cooling
    )

    return best_cost  

if __name__ == "__main__":
    study = optuna.create_study(direction="minimize")
    
    study.optimize(objective, n_trials=30)

    print("\n----------------------------------")
    print("Best params found:")
    for key, value in study.best_params.items():
        print(f"  {key}: {value}")
    print(f"Best cost: {study.best_value}")
```
