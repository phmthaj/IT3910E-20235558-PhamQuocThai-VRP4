```python
import optuna
from alns_loggi import parse_vrp_file, alns  # giả sử bạn có file alns_loggi.py chứa 2 hàm này

def objective(trial):
    reward_improve = trial.suggest_float("reward_improve", 1.0, 10.0)
    reward_accept = trial.suggest_float("reward_accept", 0.1, 5.0)
    weight_decay = trial.suggest_float("weight_decay", 0.5, 0.99)
    weight_learning = trial.suggest_float("weight_learning", 0.01, 0.5)
    cooling = trial.suggest_float("cooling", 0.9, 0.999)
    T0_factor = trial.suggest_float("T0_factor", 0.001, 0.1)

    instance_path = "data/instances/Loggi-n401-k23.vrp"
    data = parse_vrp_file(instance_path)

    dist = data["dist"]
    depot = data["depot"]
    demands = data["demands"]
    cap = data["capacity"]

    routes, best_cost = alns(
        dist, depot, demands, cap,
        iters=200,      
        seed=17,
        reward_improve=reward_improve,
        reward_accept=reward_accept,
        weight_decay=weight_decay,
        weight_learning=weight_learning,
        weight_update_period=50,
        T0_factor=T0_factor,
        cooling=cooling
    )

    return best_cost 

if __name__ == "__main__":
    study = optuna.create_study(direction="minimize")
    study.optimize(objective, n_trials=50)

    print("Best params:", study.best_params)
    print("Best cost:", study.best_value)
```
